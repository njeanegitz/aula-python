```python
import numpy as np
```

```python
array_1 = np.array([10,20,30,40,50])
array_1
```

```python
#função len mostra a quantidade de elementos do array
contador = 0
while contador < len(array_1):
    contador += 1
    print(contador)
```

```python
# for pega elemnto por elemento ate acabar
for i in range(len(array_1)):
    print(array_1[i])
```

```python
contador = 0
for _ in range(len(array_1)):
    print(array_1[contador])
    contador += 1
```

```python
# enumerate é uma função que retorna o índice e o valor do elemento do array
for i in enumerate(array_1):
    print(i)
```

```python
for key, value in enumerate(array_1):
    print(key, value)
```
