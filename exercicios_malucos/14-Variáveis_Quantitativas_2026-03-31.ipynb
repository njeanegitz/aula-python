## 01 – Usando o método np.random.normal(), gere valores de expressão gênica de um conjunto de 1.000 genes. Esse conjunto de ter valores up e downregulated.

```python
import numpy as np
```

```python
expressao = np.random.normal(loc=0, scale=1, size=1000)
expressao
```

## 02 – Usando o método np.random.randint(), gere valores para o número de genes diferencialmente expressos em um conjunto de 1.000 pacientes. Esse conjunto deve ter valores up e downregulated.

Instancie um objeto da classe DataFrame, com os dados gerados nos dois exercícios.

```python
import pandas as pd
```

```python
dif_expressos = np.random.randint (1, 100, size=1000)
dif_expressos
```

```python
df_exp = pd.DataFrame({
    "Expressão_Gênica": expressao,
    "Genes_Diferenciais": dif_expressos
})

print(df_exp.head())
```

## 03 – Categorize em grupos de 6 os dados contidos na DataFrame anterior, segundo critérios escolhidos por você. Explique os critérios.

R.: Dividi os dados em 6 grupos de acordo com o valor da expressão gênica: muito downregulated, moderado downregulated, leve downregulated, leve upregulated, moderado upregulated e muito upregulated.

```python
bins = [-np.inf, -1.0, -0.5, 0, 0.5, 1.0, np.inf]
labels = ["Muito Down", "Moderado Down", "Leve Down", 
          "Leve Up", "Moderado Up", "Muito Up"]

df_exp["Categoria"] = pd.cut(df_exp["Expressão_Gênica"], bins=bins, labels=labels)
print(df_exp["Categoria"].value_counts())
```

```python
print(df_exp.head())
```

## 04 – Defina uma função em Python que receba uma pd.Series contendo uma variável quantitativa e retorne uma DataFrame contendo a tabela de frequência.

Obtenha a tabela de frequência das variáveis criadas nos exercícios 1 e 2.

```python
def tabela_frequencia(serie: pd.Series) -> pd.DataFrame:
    freq_abs = serie.value_counts().sort_index()
    freq_rel = serie.value_counts(normalize=True).sort_index()
    tabela = pd.DataFrame({
        "Frequência Absoluta": freq_abs,
        "Frequência Relativa (%)": (freq_rel * 100).round(2)
    })
    return tabela
```

```python
freq_exp = tabela_frequencia(df_exp["Expressão_Gênica"])
print(freq_exp.head())

freq_diff = tabela_frequencia(df_exp["Genes_Diferenciais"])
print(freq_diff.head())
```

## 05 – Gere os gráficos de barras e linhas para as frequências em cada umas das variáveis.

```python
import matplotlib.pyplot as plt
```

```python
freq_exp = df_exp["Categoria"].value_counts()
freq_num = df_exp["Genes_Diferenciais"].value_counts().sort_index()
```

### Expressão

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
freq_exp.plot(kind="bar", color="skyblue")
plt.title("Frequência - Expressão (Barras)")
plt.xlabel("Categorias")
plt.ylabel("Frequência")
plt.tight_layout()
plt.show()
```

```python
plt.subplot(1,2,2)
freq_exp.plot(kind="line", marker="o", color="orange")
plt.title("Frequência - Expressão (Linhas)")
plt.xlabel("Categorias")
plt.ylabel("Frequência")
plt.tight_layout()
plt.show()
```

### Número

```python
plt.figure(figsize=(25,5))

plt.subplot(1,2,1)
freq_num.plot(kind="bar", color="lightgreen")
plt.title("Frequência - Número (Barras)")
plt.xlabel("Genes Diferenciais")
plt.ylabel("Frequência")
plt.tight_layout()
plt.show()
```

```python
plt.subplot(1,2,2)
freq_num.plot(kind="line", marker="o", color="red")
plt.title("Frequência - Número (Linhas)")
plt.xlabel("Genes Diferenciais")
plt.ylabel("Frequência")
plt.tight_layout()
plt.show()
```

## 06 – Gere os histogramas para as variáveis obtidas nos exercícios 1 e 2.

```python
#Histograma de expressão
plt.figure(figsize=(12,5))
plt.hist(df_exp["Expressão_Gênica"], bins=30, color="skyblue", edgecolor="black")
plt.title("Histograma - Expressão Gênica")
plt.xlabel("Valor de Expressão")
plt.ylabel("Frequência")
plt.show()
```

## 07 – Gere um gráfico de dispersão para as variáveis obtidas nos exercícios 1 e 2.

```python
plt.figure(figsize=(8,6))
plt.scatter(df_exp["Expressão_Gênica"], df_exp["Genes_Diferenciais"], 
            alpha=0.6, color="purple", edgecolors="black")

plt.title("Gráfico de Dispersão - Expressão vs Número")
plt.xlabel("Expressão Gênica")
plt.ylabel("Número de Genes Diferenciais")
plt.show()
```

## 08 – Gere o mesmo gráfico do exercício 06, porém, categorizado.

```python
freq_exp_cat = df_exp["Categoria"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(freq_exp_cat.index, freq_exp_cat.values, color="skyblue", edgecolor="black")
plt.title("Histograma - Expressão Gênica (Categorizada)")
plt.xlabel("Categorias")
plt.ylabel("Frequência")
plt.show()
```

```python
bins_num = [0, 20, 40, 60, 80, 100]
labels_num = ["0-20", "21-40", "41-60", "61-80", "81-100"]

df_exp["Categoria_Numero"] = pd.cut(df_exp["Genes_Diferenciais"], bins=bins_num, labels=labels_num)

freq_num_cat = df_exp["Categoria_Numero"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(freq_num_cat.index.astype(str), freq_num_cat.values, color="lightgreen", edgecolor="black")
plt.title("Histograma - Número de Genes Diferenciais (Categorizado)")
plt.xlabel("Intervalos")
plt.ylabel("Frequência")
plt.show()
```

## 09 – Gere o mesmo gráfico do exercício 07, porém, categorizado.

```python
plt.figure(figsize=(10,7))

for categoria, cor in zip(df_exp["Categoria"].unique(),
                          ["blue","lightblue","cyan","orange","red","darkred"]):
    subset = df_exp[df_exp["Categoria"] == categoria]
    plt.scatter(subset["Expressão_Gênica"], subset["Genes_Diferenciais"],
                label=categoria, alpha=0.6, edgecolors="black", color=cor)

plt.title("Gráfico de Dispersão - Expressão vs Número (Categorizado)")
plt.xlabel("Expressão Gênica")
plt.ylabel("Número de Genes Diferenciais")
plt.legend(title="Categoria")
plt.show()
```
