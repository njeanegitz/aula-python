## 01 – Descreva o racional da sua resposta:

R.: Após instanciar a lista com 5.000 inteiros, para cada valor divir em 2 e analisar o resto da divisão. Se o resto for = 0, então o número é par e se for > 0, então é ímpar. Após isso, somar quantos são pares e quantos são ímpares.

## 02 – Desenhe o fluxograma da resposta:

## 03 – Escreva o programa em Python:

```python
lista = list(range(5000))
pares = 0
impares = 0
```

```python
for numero in lista:
    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1
print(f"Números pares: {pares}")
print(f"Números ímpares: {impares}")
```

## 06 – Desenhe o fluxograma da resposta:

## 07 – Escreva o programa em Python:

```python
import pandas as pd
```

```python
df_genes = pd.read_csv("df_genes.csv", index_col=0, header=0)
df_genes
```

```python
pares = 0
impares = 0
for valor in df_genes["caracteres"]:
    if valor % 2 == 0:
        pares += 1
    else:
        impares += 1
print(f"Números pares: {pares}")
print(f"Números ímpares: {impares}")
```

## 08 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
par = (df_genes["caracteres"] % 2 == 0).sum()
impar = (df_genes["caracteres"] % 2 != 0).sum()
print(f"Números pares: {pares}")
print(f"Números ímpares: {impares}")
```
