```python
def contar_inteiros(object: list) -> int:
    contador = 0
    for x in range(len(object)):
        if type(object[x]) == int:
            contador += 1
    return contador
```

```python
lista = [1, 2, 3, 'a', 'b', 4.5, 5]
resultado = contar_inteiros(lista)
print(f'O número de inteiros na lista é: {resultado}')
```
