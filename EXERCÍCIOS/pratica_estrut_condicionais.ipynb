```python
import numpy as np
import pandas as pd
```

```python
array = np.random.normal(20, 2.5, 40).reshape(8,5)
array
```

```python
if array[3,1] >= 200:
    resposta = "O valor é maior ou igual a 200"
else:
    resposta = "O valor não é maior ou igual a 200"
resposta
```

```python
# elif significa que se a primeira condição for falsa, ele vai testar a segunda condição, e se a segunda condição for falsa, ele vai testar a terceira condição, e assim por diante. Se todas as condições forem falsas, ele vai executar o código do else.
if array[3,1] >= 200:
    resposta = "O valor é maior ou igual a 200"
elif array[3,1] < 20:
    resposta = "O valor é maior ou igual a 20"
else:
    resposta = "O valor não é maior ou igual a 200"
resposta
```
