## 10 – Importe o data.csv (dados de seguro na Flórida em 2012) para uma DataFrame. Armazene em um dicionário o número de entradas, uma lista com as classes das colunas e outra lista com o número de elementos de cada classe, três listas, contendo a média, o desvio padrão e o terceiro quartil da primeira, quarta e nona colunas . Recupere o dicionário.

```python
df_tabela = pd.read_csv("data.csv")
df_tabela
```

```python
#Numero de entradas
entradas = df_tabela.shape[0]
entradas
```

```python
#Lista com as classes das colunas
classes_colunas = list(df_tabela.dtypes)
classes_colunas
```

```python
#Lista com o nº de elementos de cada classe
classes_elementos = df_tabela.dtypes.value_counts()
classes_elementos
```

```python
classes_elementos = list(df_tabela.dtypes.value_counts())
classes_elementos
```

```python
#Lista média da primeira, quarta e nona coluna
media = df_tabela.iloc[:, [0, 3, 8]]
media = media.mean(axis=1)
media
```

```python
media = list(df_tabela.iloc[:, [0, 3, 8]].mean(axis=1))
media
```

```python
#Lista desvio padrão
desvio_padrao = list(df_tabela.iloc[:, [0, 3, 8]].std(axis=0))
desvio_padrao
```

```python
#Lista com o terceiro quartil da primeira, quarta e nona colunas
quartil = list(df_tabela.iloc[:, [0, 3, 8]].quantile(0.75))
quartil
```

```python
#Dicionário 
dicionario_tabela = {
    "entradas": entradas,
    "classes_colunas": classes_colunas,
    "classes_elementos": classes_elementos,
    "media": media,
    "desvio_padrao": desvio_padrao,
    "quartil": quartil
}
dicionario_tabela
```
